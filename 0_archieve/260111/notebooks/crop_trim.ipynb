```python
import pandas as pd

def pretty(df):
    # 소수점 뒤 불필요한 0 제거 (출력 전용)
    def smart_float(x):
        if isinstance(x, float):
            s = format(x, 'f')
            return s.rstrip('0').rstrip('.')
        return x

    return (
        df.style
        .format(smart_float)  # float 보기 좋게
        .set_table_styles(
            [
                {"selector": "table", "props": [("border-collapse", "collapse")]},
                {"selector": "th", "props": [("border", "1px solid #444"), ("background", "#fafafa"), ("padding", "6px")]},
                {"selector": "td", "props": [("border", "1px solid #ccc"), ("padding", "6px")]},
            ]
        )
        .set_properties(**{"text-align": "center"})
    )


pd.set_option("styler.render.max_columns", 200)
pd.set_option("styler.render.max_rows", 200)

pd.options.display.float_format = lambda x: ('%f' % x).rstrip('0').rstrip('.')
```

```python
import yaml

yaml_path = r"/home/user/문서/workspace/python/yaml/"
yaml_name = "crop_trim.yaml"

with open(f"{yaml_path}{yaml_name}", "r", encoding="utf-8") as f:
    yaml_data = yaml.safe_load(f)

print(yaml_data)
```

```python
import chardet


raw_folder = "/home/user/문서/raw_data/"

focus_item = "crop_price_korea"

file = f"{raw_folder}{yaml_data[focus_item]['raw_name']}"
    
with open(file, "rb") as f:
    enc = chardet.detect(f.read())['encoding']
    
df = pd.read_csv(file, encoding=enc, engine="python")



pretty(df.head())
```

```python
df= df.melt(
    id_vars=["품목별","규격별", "항목", "단위"],   # 그대로 유지할 컬럼
    var_name="시점",             # 기존 열 이름을 넣을 새 컬럼명
    value_name="데이터"           # 값이 들어갈 새 컬럼명
)

pretty(df.head())
```

```python
df[["연도", "월", "일"]] = df["시점"].str.split(". ", expand=True)
pretty(df.head())
```

```python
df = df[[ "연도", "월", "일" ] + [c for c in df.columns if c not in ["시점", "연도", "월", "일"]]]

df["연도"] = df["연도"].astype("int16")
df["월"] = df["월"].astype("int8")
df["일"] = df["일"].astype("int8")



pretty(df.head())
```

```python
df["데이터"] = pd.to_numeric(df["데이터"], errors="coerce").astype("float32")

pretty(df.head())
```

```python
df = df.convert_dtypes()  # 기본 자동 변환

df["연도"] = df["연도"].astype("int16")
df["월"] = df["월"].astype("int8")
df["일"] = df["일"].astype("int8")
df["품목별"] = df["품목별"].astype("category")
df["규격별"] = df["규격별"].astype("category")
df["단위"] = df["단위"].astype("category")
df["항목"] = df["항목"].astype("category")
df["데이터"] = df["데이터"].astype("float32")  # 숫자라면 float32가 최적

data_folder = "/home/user/gdrive/data/"

df.to_parquet(f"{data_folder}{yaml_data[focus_item]['parquet_name']}", index=False)
```
